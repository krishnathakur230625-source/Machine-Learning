[1,2,3,4,5] → These are your x-values.

[1,4,9,16,25] → These are your y-values (squares of x-values).

'r^-' → This is a format string that styles the plot:

'r' → Red color

'^' → Triangle-up markers for each point

'-' → Solid line connecting the points

```python
plt.grid(True)
plt.title('squares of the numbers')
plt.xlabel('The Numbers', fontsize=18, color='red')
plt.ylabel('The squares',fontsize=18, color='red')
plt.plot([1,2,3,4,5], [1,4,9,16,25], 'r^--', linewidth=3.0)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a sample dataset
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Revenue': [15000, 18000, 12000, 22000, 25000, 21000],
    'Expenses': [10000, 12000, 11000, 13000, 14000, 13000]
}

# Save to a CSV file
df_sample = pd.DataFrame(data)
df_sample.to_csv('monthly_report.csv', index=False)
print('Dataset created and saved as monthly_report.csv')

# Display the first few rows
display(df_sample.head())
```

```python
# Load the dataset we just created
df = pd.read_csv('monthly_report.csv')

# Create a comparison plot
plt.figure(figsize=(10, 6))
plt.plot(df['Month'], df['Revenue'], marker='o', label='Revenue', color='green', linewidth=2)
plt.plot(df['Month'], df['Expenses'], marker='s', label='Expenses', color='red', linestyle='--')

plt.title('Monthly Revenue vs Expenses')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Prepare data
x_numeric = np.arange(len(df['Month']))
y_revenue = df['Revenue']
z_expenses = df['Expenses']

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Create a 3D scatter plot
ax.scatter(x_numeric, y_revenue, z_expenses, c='blue', marker='o', s=100, label='Data Points')

# Create a 3D line plot connecting them
ax.plot(x_numeric, y_revenue, z_expenses, color='gray', linestyle='--', alpha=0.5)

# Customizing labels
ax.set_xticks(x_numeric)
ax.set_xticklabels(df['Month'])
ax.set_xlabel('Month')
ax.set_ylabel('Revenue ($)')
ax.set_zlabel('Expenses ($)')
ax.set_title('3D View of Financial Performance')

plt.legend()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate correlation matrix for numerical columns
correlation_matrix = df[['Revenue', 'Expenses']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap: Revenue vs Expenses')
plt.show()
```

```python
from matplotlib import cm

# Prepare data for surface plot
x_range = np.arange(len(df['Month']))
y_revenue = df['Revenue'].values
X, Y = np.meshgrid(x_range, y_revenue)
# Create a dummy Z for surface visualization based on relationship
Z = np.tile(df['Expenses'].values, (len(y_revenue), 1))

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap=cm.viridis, linewidth=0, antialiased=False, alpha=0.8)

# Customization
ax.set_xticks(x_range)
ax.set_xticklabels(df['Month'])
ax.set_xlabel('Month')
ax.set_ylabel('Revenue ($)')
ax.set_zlabel('Expenses ($)')
ax.set_title('3D Surface Plot of Financial Data')

fig.colorbar(surf, shrink=0.5, aspect=5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


np.random.seed(42)
x = np.random.randint(0, 10, 100)
y = np.random.randint(0, 10, 100)


hist, xedges, yedges = np.histogram2d(x, y, bins=10)

xpos, ypos = np.meshgrid(
    xedges[:-1] + 0.25,
    yedges[:-1] + 0.25,
    indexing="ij"
)

xpos = xpos.ravel()
ypos = ypos.ravel()
zpos = np.zeros_like(xpos)


dx = dy = 0.5 * np.ones_like(zpos)
dz = hist.ravel()

fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

ax.bar3d(xpos, ypos, zpos, dx, dy, dz, shade=True)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Frequency')

plt.title("3D Histogram")
plt.show()
```

### Interactive 3D Bubble Chart (Portable)
This cell uses **Plotly** to create a highly interactive 3D visualization.
- **Interactivity:** You can rotate, zoom, and hover.
- **Portability:** The `fig.write_html` command saves the entire plot as an interactive file that can be opened in any browser or shared.

```python
import plotly.graph_objects as go
import pandas as pd

df = pd.read_csv('monthly_report.csv')

fig = go.Figure(data=[go.Scatter3d(
    x=df['Month'],
    y=df['Revenue'],
    z=df['Expenses'],
    mode='markers',
    name='Financial Data',
    marker=dict(
        size=df['Revenue']/1000,
        color=df['Expenses'],
        colorscale='Viridis',
        opacity=0.8,
        showscale=True,
        colorbar=dict(title='Expense Level')
    ),
    text=df['Month'],
)])

fig.update_layout(
    title='Interactive Monthly Financial Analysis',
    # Move legend to the top
    legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1),
    showlegend=True,
    scene=dict(
        xaxis_title='Month',
        yaxis_title='Revenue ($)',
        zaxis_title='Expenses ($)',
        bgcolor='rgb(20, 20, 20)'
    ),
    margin=dict(l=0, r=0, b=0, t=40)
)

fig.show()

fig.write_html('interactive_3d_report.html')
print("Plot saved as 'interactive_3d_report.html'.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- CUSTOMIZE YOUR METRICS HERE ---
x_metric = 'Month'
y_metric = 'Revenue'
z_metric = 'Expenses'
color_metric = 'Revenue'  # This defines the heatmap
# ----------------------------------

x_data = np.arange(len(df[x_metric]))
y_data = df[y_metric]
z_data = df[z_metric]
c_data = df[color_metric]

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot using the custom metrics
scatter = ax.scatter(x_data, y_data, z_data, c=c_data, cmap='magma', s=120, edgecolors='w', alpha=0.8)

# Dynamic labeling based on your choices
ax.set_xticks(x_data)
ax.set_xticklabels(df[x_metric])
ax.set_xlabel(x_metric)
ax.set_ylabel(f'{y_metric} ($)')
ax.set_zlabel(f'{z_metric} ($)')
ax.set_title(f'Custom 3D Analysis: {y_metric} vs {z_metric}')

# Colorbar for the custom heatmap metric
cbar = fig.colorbar(scatter, ax=ax, shrink=0.5, aspect=10)
cbar.set_label(f'Intensity: {color_metric}')

plt.show()
```
